```python
# rename column names
df = df.rename(columns= {'Country Name' : 'Country Name',
                        'Country Code': 'Country Code',
                        'Series Name' : 'Economic Indicator',
                        'Series Code' : 'Indicator Code' })
```

```python
#clean years
df.columns = [ re.sub(r"\s*\[YR\d+\]", "", col) if re.search(r"\d{4}\s*\[YR\d+\]", col) else col for col in df.columns]
```
